# GhostChef AI – Demand Forecasting & Inventory Optimization
## Day 1 Part 4 — Data Understanding

````python
import io
import sys

MD_FILE = "output.md"
md_lines = []

def mdprint(*args, **kwargs):
    text = ' '.join(str(a) for a in args)
    md_lines.append(text)

def mdheading(text, level=2):
    md_lines.append('#' * level + ' ' + text)
    md_lines.append('')

def mdcode(text):
    md_lines.append('```')
    md_lines.append(text)
    md_lines.append('```')
    md_lines.append('')

def df_to_md(df, max_rows=5):
    md_lines.append(df.head(max_rows).to_markdown())
    md_lines.append('')

def df_info_to_md(df, name):
    buf = io.StringIO()
    df.info(buf=buf)
    info_str = buf.getvalue()
    md_lines.append(f'**{name} Info:**')
    md_lines.append('')
    mdcode(info_str)

def save_md():
    with open(MD_FILE, 'w', encoding='utf-8') as f:
        f.write('\n'.join(md_lines))
    print(f'Markdown report saved to {MD_FILE}')

mdheading('GhostChef AI - Data Exploration Report', 1)
````

````python
import pandas as pd
import os
````

### Load Datasets

````python
sales = pd.read_csv("../data/raw/sales_train_validation.csv")
mdprint('sales loaded')

calendar = pd.read_csv("../data/raw/calendar.csv")
mdprint('calendar loaded')

prices = pd.read_csv("../data/raw/sell_prices.csv")
mdprint('prices loaded')
````

### Data Shapes

````python
mdheading('Data Shapes')
mdprint(f'Sales: {sales.shape}')
mdprint(f'Calendar: {calendar.shape}')
mdprint(f'Prices: {prices.shape}')
````

### Column Counts

````python
mdheading('Column Counts')
mdprint(f'sales columns: {sales.shape[1]}')
mdprint(f'calendar columns: {calendar.shape[1]}')
mdprint(f'prices columns: {prices.shape[1]}')
````

### Head — First 5 Rows

````python
mdheading('Sales Head')
df_to_md(sales)

mdheading('Calendar Head')
df_to_md(calendar)

mdheading('Prices Head')
df_to_md(prices)
````

### Info — Schema & Memory

````python
df_info_to_md(sales, 'Sales')
df_info_to_md(calendar, 'Calendar')
df_info_to_md(prices, 'Prices')
````

### Null Counts Per Column

````python
mdheading('Null Counts')

for name, df in [('Sales', sales), ('Calendar', calendar), ('Prices', prices)]:
    nulls = df.isnull().sum()
    nulls_nonzero = nulls[nulls > 0]
    if len(nulls_nonzero) > 0:
        md_lines.append(f'**{name} — Columns with nulls:**')
        md_lines.append('')
        md_lines.append(nulls_nonzero.to_markdown())
        md_lines.append('')
    else:
        mdprint(f'{name}: No nulls found')
````

### Describe — Numerical Summaries

````python
mdheading('Numerical Summaries')

for name, df in [('Sales', sales), ('Calendar', calendar), ('Prices', prices)]:
    mdheading(f'{name} — describe()', level=3)
    md_lines.append(df.describe().to_markdown())
    md_lines.append('')
````

### Unique Value Counts — Categorical Columns

````python
mdheading('Unique Value Counts — Categorical Columns')

for name, df in [('Sales', sales), ('Calendar', calendar), ('Prices', prices)]:
    cat_cols = df.select_dtypes(include='object').columns
    if len(cat_cols) > 0:
        mdheading(f'{name} — Categorical Columns', level=3)
        for col in cat_cols:
            mdprint(f'{col}: {df[col].nunique()} unique values')
````

### Save Markdown Report

````python
save_md()
````
